# Day6 Functions 
A function in Python is a reusable block of code that executes a specific task only when explicitly called. Functions help break large programs into smaller, organized, and manageable chunks while completely eliminating code repetition.

### Basics practice 

```python
def my_function():
    print("Welcome to Compuational cosmology")
my_function()
my_function()
```

```python
def name():
    print("MUuskan jain")
name()
```

```python
# sum of different numbers
def sum(a,b):
    return a + b

result = sum(8,9)
print(result)
```

```python
# square of number 
def square(a):
    return a**2
number_square = square(5)
print(number_square)
```

```python
def check_even_odd(number):
    if number % 2 == 0:
        print("The number is even.")
    else:
        print("The number is odd.")

check_even_odd(8)
check_even_odd(5)
```

# Physics based functions

```python
# Force calcualtor 
def force(mass,acceleration):
    return mass*acceleration
calculate_force = force(8,9)
print(calculate_force)
```

```python
# calculating momentum 
def momentum(mass,velocity):
    return mass*velocity
calculate_momentum = momentum(9.7,8)
print(calculate_momentum)
```

```python
# celsius to kelvin
def temp(celsius):
    return celsius + 273.15
final_temp = temp(78)
print(final_temp)
```

# MINI PROJECT

```python
mass = float(input("The mass of object",)) # input mass 
velocity = float(input("The velocity of object",)) #input velocity
acceleration = float(input("The acceleration of obejct",)) #input acceleration

# printing input value
print("object mass:",mass) # kg
print("object velocity:",velocity) # m/s
print("object acceleration:",acceleration) # m/s^2

# calculate force
def force(mass,acceleration):
    return mass*acceleration

# calculate momentum
def momentum(mass,velocity):
    return mass*velocity

# calculate kinetic energy
def kinetic_energy(mass,velocity):
    return 0.5*mass*velocity**2

#calling functions
final_force = force(mass,acceleration)
final_momentum = momentum(mass,velocity)
final_kinetic_energy = kinetic_energy(mass,velocity)

print("object force :",f"{final_force:.2f}","N")
print("obejct momentum:",f"{final_momentum:.2f}","kg*m/s")
print("object kinetic energy:",f"{final_kinetic_energy:.2f}","joule")
```
