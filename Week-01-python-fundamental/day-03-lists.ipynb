### In Python, a list is a built-in, mutable, and ordered collection data type designed to store multiple items in a single variable. Items are enclosed in square brackets [] and separated by commas.

```python
student_profile_1= [
    "Muskan jain", 
    "cosmologist", 
    "21", 
    "bachelor's degree in physics"]
print(student_profile_1) 
print("Name of student:" + student_profile_1[0]) # prints the first element of the list
print("Aim of student:" + student_profile_1[1]) # prints the second element of the list
print("Age of student:" + student_profile_1[2]) # prints the third element of the list 
print("Education of student:" + student_profile_1[3]) # prints the fourth element of the list   

student_profile_1.append("mdu, rohtak")
print(student_profile_1) # prints the list after adding a new element to it
student_profile_1.remove("21")
print(student_profile_1) # prints the list after removing an element from it
student_profile_1.insert(3,"23") # inserts a new element at the specified index
print(student_profile_1) # prints the list after inserting a new element
```

### Adding elements in lists

```python
from abc import ABC, abstractmethod

class StudentProfileBase(ABC):
    def __init__(self, profile):
        self.profile = list(profile)

    @abstractmethod
    def add_profile(self, item):
        """Add an item to the profile."""
        pass

    @abstractmethod
    def remove_profile(self, item):
        """Remove an item from the profile if it exists."""
        pass

    @abstractmethod
    def insert_profile(self, index, item):
        """Insert an item at a given index."""
        pass

existing_profile = student_profile_1

class student_profile_1(StudentProfileBase):
    def add_profile(self, item):
        self.profile.append(item)
        return self.profile

    def remove_profile(self, item):
        if item in self.profile:
            self.profile.remove(item)
        return self.profile

    def insert_profile(self, index, item):
        self.profile.insert(index, item)
        return self.profile

    def append(self, item):
        return self.add_profile(item)

    def remove(self, item):
        return self.remove_profile(item)

    def insert(self, index, item):
        return self.insert_profile(index, item)

    def __getitem__(self, index):
        return self.profile[index]

    def __iter__(self):
        return iter(self.profile)

    def __len__(self):
        return len(self.profile)

    def __repr__(self):
        return repr(self.profile)

    def __str__(self):
        return str(self.profile)

student_profile_1 = student_profile_1(existing_profile)
```
