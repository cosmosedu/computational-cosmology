**Calculate the area of a circle**

```python
radius = 6.7
area = 3.14 * radius **2
print("The area of the circle is :", area)
circumfernece = 2 * 3.14 * radius 
print("The circumference of the circle is :", circumfernece)
```

**standard method to solve area of circle**

```python
import math
radius = float(input("Enter the radius of the circle: "))
area = math.pi * radius ** 2
circumference = 2 * math.pi * radius
print("The area of the circle is:", f"{area:.2f}")
print("The circumference of the circle is:", f"{circumference:.2f}")
```
