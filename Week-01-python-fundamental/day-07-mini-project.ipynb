```python
time = [3,6,9,12,15,18,21,24]
velocity = [5, 10, 15, 20, 25, 30, 35,40]

def average_velocity(velocity):
    return sum(velocity) / len(velocity)

velocity_final = average_velocity(velocity)
print("The average velocity is:", velocity_final)

distance = 0
for i in range(len(time)):
    distance += time[i] * velocity[i]
print("The total distance is:", distance)

import matplotlib.pyplot as plt
plt.plot(time, velocity)
plt.xlabel("Time (s)")
plt.ylabel("Velocity (m/s)")
plt.title("Velocity vs Time")
plt.show()
```
