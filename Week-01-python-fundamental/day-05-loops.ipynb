### NUMBERS FROM 1 TO 20 USING FOR LOOP 

```python
# we use range inbuilt function to perform this task
for i in range(1,21):
    print(i)
```

```python
# print the first 15 even numbers using while loop
i = 1
while i <= 15:
    print(i*2)
    i += 1  
```

```python
# table of 6 using for loop
for i in range(1,11):
    print(6*i)
```

```python
# sum mumber from 1 to 10 using while loop
i = 1
sum = 0
while i <= 10:
    sum += i
    i += 1
print("Sum of numbers from 1 to 10 is:", sum)
```

# physics based probelms 

```python
# distance travelled
speed = int(input("enter the speed travelled in m/s: ")) # speed input in m/s
time = 1 # starting time in seconds 
while time <=10:
    distance = speed * time
    print("Distance travelled in", time, "seconds is:", distance, "meters")
    time += 1
```

# free fall 

```python
for i in range(1, 11):
    displacement = 0.5 * 9.8 * i**2
    print("Displacement after", i, "seconds is:", f"{displacement:.2f}","meters")
```

# Rocket fuel simulation 

```python
fuel = 100 # initial fuel level in percentage
minute = 0
while fuel > 0:
    minute += 1
    print("Minute:", minute, "Fuel level:", fuel, "%")
    fuel -= 10 # decrease fuel level by 10% each iteration  
print("Fuel level is empty. Stopping the engine.")
```

# Temperature conversion table 

```python
for i in range(0, 100,10): # tempreature in celsius from 0 to 100 with step of 10
    temperature = i + 273.15 # convert celsius to kelvin
    print( "convert celsius to kelvin:", i, "°C =", temperature, "K")
```
