**Task 1** : **Calculating Kinetic Energy and momentum by taking "INPUT"**

```python
mass = float(input("mass of the object in kilograms: "))
print("The mass of the object is :", mass,"kilograms")
velocity = float(input("Velocity of the object in meters per second: "))
print("The velocity of the objects is :", velocity,"meters per second")
kinetic_energy = 0.5 * mass * velocity ** 2
momentum = mass * velocity
print("The momentum of the object is:", f"{momentum:.2f}", "kg*m/s")
print("The kinetic energy of the object is:", f"{kinetic_energy:.2f}", "Joules")
```

**Task2** : **Calcualting Force**

```python
mass = float(input("mass of the object in kilograms: "))
print("The mass of the object is :", mass,"kilograms")
acceleration = float(input("Acceleration of the object in meters per second squared: "))
print("The acceleration of the objects is :", acceleration,"meters per second squared")
force = mass * acceleration
print("The force acting on the object is:", f"{force:.2f}", "Newtons")   
```

**Task 3 : Density**

```python
mass = float(input("mass of the object in kilograms: "))
print("The mass of the object is :", mass,"kilograms")
volume = float(input("Volume of the object in cubic meters: "))
print("The volume of the objects is :", volume,"cubic meters")
density = mass / volume
print("The density of the object is:", f"{density:.2f}","kg/m^3")  
```
