# DAY 4 IF , ELIF, ELSE STATEMENTS 

```python
velocity = int(input("enter velocity: ")) # in km/h
if velocity > 50:
    print("You are driving too fast!")
else:
    print("You are driving within the speed limit.")
```

```python
stars_temperature = int(input("Enter the temperature of the star in Kelvin: "))
print("The temperature of the star is:", stars_temperature, "K")
if stars_temperature > 9000:
    print("Star is hot")
elif stars_temperature > 4000:
    print("Star is warm")
else:
    print("Star is cold")
```

## Star temperature classification

```python
star_temp = int(input("Enter the temperature of the star in Kelvin: "))
if star_temp >= 30000:
    print("o-type star")
elif star_temp >= 10000:
    print("B-type star")
elif star_temp >= 7500:
    print("A-type star")
elif star_temp >= 6000:
    print("F-type star")
elif star_temp >= 5200:
    print("G-type star")
elif star_temp >= 3700:
    print("K-type star")    
else:
    print("M-type star")
```
