# Day 3 Numpy Mathematical & statistical operations

### Part 1 : Mathematical functions

```python
import numpy as np
a = np.array([7,8,5,3,5,4])
b = np.array([[3,5],[7,8]])

# square root of a and b arrays
print("Square root of a:", np.sqrt(a))
print("Square root of b:")
print(np.sqrt(b))
print()

# square of a and b arrays
print("Square of a:", np.square(a))
print("Square of b:")
print(np.square(b))
print()

#exponential of a and b arrays
print("Exponential of a:", np.exp(a))
print("Exponential of b:",np.exp(b))
print()

#logarithm of a and b arrays
print("Logarithm of a:", np.log(a))
print("Logarithm of b:")
print(np.log(b))
print()

#sine of a and b arrays
print("Sine of a:", np.sin(a))
print("Sine of b:")
print(np.sin(b))
print()
```
