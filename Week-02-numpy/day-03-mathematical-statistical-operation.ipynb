## Part1 mathematical function 

```python
import numpy as np

a = np.array([7, 8, 5, 3, 5, 4])

b = np.array([
    [3, 5],
    [7, 8]
])

print("Square root of a:", np.sqrt(a))
print()
print("Square root of b:")
print(np.sqrt(b))
print()

print("Square of a:", np.square(a))
print()
print("Square of b:")
print(np.square(b))
print()

print("Exponential of a:", np.exp(a))
print()
print("Exponential of b:")
print(np.exp(b))
print()

print("Natural log of a:", np.log(a))
print()
print("Natural log of b:")
print(np.log(b))
print()

print("Sine of a:", np.sin(a))
print()
print("Sine of b:")
print(np.sin(b))
print()

print("Cosine of a:", np.cos(a))
print()
print("Cosine of b:")
print(np.cos(b))
```

## Part 2 — Basic Statistical Operations

```python
data = np.array([10, 20, 30, 40, 50])

print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Standard deviation:", np.std(data))
print("Variance:", np.var(data))
print("Minimum:", np.min(data))
print("Maximum:", np.max(data))
print("Sum:", np.sum(data))
```

## Part 3 — Mean vs Median

```python
temp = np.array([10, 11, 12, 13, 14, 100])

print("Mean of temp:", np.mean(temp))
print("Median of temp:", np.median(temp))
```

The extreme value 100 strongly affects the mean, while the median is much less affected.

## Part 4 — Understanding axis

```python
data = np.array([
    [5000, 5100, 5050],
    [6000, 6100, 6050],
    [7000, 7100, 7050],
    [8000, 8100, 8050]
])

print("Data:")
print(data)

print()
print("Mean of entire dataset:")
print(np.mean(data))

print()
print("Mean along axis=0:")
print(np.mean(data, axis=0))

print()
print("Mean along axis=1:")
print(np.mean(data, axis=1))
```

axis=0 → column-wise operation

axis=1 → row-wise operation

## Part 5 — Astronomy: Stellar Temperatures

```python
temperature = np.array([
    3200, 3500, 4100, 4500, 5200,
    5800, 6200, 7000, 8500, 10000
])

print("Mean temperature:", np.mean(temperature))
print("Median temperature:", np.median(temperature))
print("Standard deviation:", np.std(temperature))
print("Variance:", np.var(temperature))
print("Minimum temperature:", np.min(temperature))
print("Maximum temperature:", np.max(temperature))
```

## Part 6 — Combine Day 2 + Day 3

```python
hot_stars = temperature[temperature > 6000]

print("Hot stars:")
print(hot_stars)

print()
print("Mean temperature of hot stars:", np.mean(hot_stars))
print("Median temperature of hot stars:", np.median(hot_stars))
print("Standard deviation:", np.std(hot_stars))
print("Minimum:", np.min(hot_stars))
print("Maximum:", np.max(hot_stars))
```

## Part 7 — Hubble Constant Measurements

```python
H0 = np.array([
    69.8, 70.2, 69.5,
    70.1, 69.9, 70.4, 69.7
])

print("Mean H0:", np.mean(H0))
print("Median H0:", np.median(H0))
print("Standard deviation:", np.std(H0))
print("Minimum H0:", np.min(H0))
print("Maximum H0:", np.max(H0))
```
