## Week 2 — Day 4: Random Numbers & Probability

### 1. Basic random numbers

```python
import numpy as np 

#generate random numbers 

a = np.random.rand()
print(a)

b = np.random.rand(5)
print(b)

c = np.random.rand(3,3)
print(c)

#generate random integers

d = np.random.randint(5)
print(d)

e = np.random.randint(1, 10)
print(e)

f = np.random.randint(1, 10,5)
print(f)

g = np.random.randint(1, 10, size=(2, 3))
print(g)

# generate random decimals 
h = np.random.uniform()
print(h)

i = np.random.uniform(1, 10)
print(i)

j = np.random.uniform(1, 10, 5)
print(j)
print()

# Mini challenge

galaxies_ID = np.random.randint(1,1000,100)
print("total number of galaxies id:",len(galaxies_ID))
print(galaxies_ID)
print("maximum galaxy id:",np.max(galaxies_ID))
print("minimum galaxy id:",np.min(galaxies_ID))
print("mean galaxy id:",np.mean(galaxies_ID))
print("galaxies with id > 500:", len(galaxies_ID[galaxies_ID > 500]))

```

## Random Seed & Reproducibility

```python
np.random.seed(42) # Set the seed for reproducibility
sample = np.random.randint(1, 100, size=5)
print("sample:", sample)

galaxies_distance = np.random.randint(10,100, size = 10)
print("galaxies distance:", galaxies_distance)
print("maximum distance:", np.max(galaxies_distance))
print("minimum distance:", np.min(galaxies_distance))
print("mean distance:", np.mean(galaxies_distance))
```

## Uniform Distribution

```python
np.random.seed(42) # Set the seed for reproducibility
k = np.random.uniform(1, 10, 5)
print(k)

data = np.random.uniform(10, 20, 1000)
print("mean:", np.mean(data))
print("standard deviation:", np.std(data))
print("median:", np.median(data))
print("minimum:", np.min(data))
print("maximum:", np.max(data))

```

## Normal / Gaussian Distribution
This is an important one for our future cosmology work because many kinds of measurement errors and observational uncertainties are approximately modeled with a Gaussian distribution.

```python
np.random.seed(42) # Set the seed for reproducibility
# np.random.normal(loc, scale, size)
sample = np.random.normal(70, 2, 10)
print("sample:", sample)

#Generated 10 random measurements centered around 70, with a standard deviation of 2.

data = np.random.normal(70, 2, 1000)
print("mean:", np.mean(data))
print("standard deviation:", np.std(data))
print("median:", np.median(data))
print("minimum:", np.min(data))
print("maximum:", np.max(data))
print()

# Astronomy example — Hubble constant
H0 = np.random.normal(70, 2, 100)
print("Mean Hubble constant:", np.mean(H0))
print("Standard deviation of Hubble constant:", np.std(H0))
print("Median Hubble constant:", np.median(H0))
print()

sample1 = np.random.normal(100,5,500)
print("Mean of sample1:", np.mean(sample1))
print("Standard deviation of sample1:", np.std(sample1))
print("Median of sample1:", np.median(sample1))
print("Minimum of sample1:", np.min(sample1))
print("Maximum of sample1:", np.max(sample1))
print(len(sample1[(sample1>95) & (sample1<105)]))
```

## Astronomy Application — Simulating Stellar Temperatures

```python
np.random.seed(42)  # Set the seed for reproducibility
temperatures = np.random.normal(5500, 800, 1000)
print("Mean temperature:", np.mean(temperatures))
print("Median temperature:", np.median(temperatures))
print("Standard deviation of temperatures:", np.std(temperatures))
print("Minimum temperature:", np.min(temperatures))
print("Maximum temperature:", np.max(temperatures))

hot_stars = temperatures > 6000
print("Number of hot stars:", np.sum(hot_stars))
print("Mean temperature of hot stars:", np.mean(temperatures[hot_stars]))

import matplotlib.pyplot as plt
plt.hist(temperatures, bins= 100)
plt.xlabel("Stellar Temperature (K)")
plt.ylabel("Number of Stars")
plt.title("Simulated Stellar Temperature Distribution")

plt.show()
plt.show()
```
